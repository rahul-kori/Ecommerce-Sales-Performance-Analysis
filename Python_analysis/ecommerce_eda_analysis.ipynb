4. SALES TRENDS OVER TIME

```python
monthly_sales = sales.groupby('order_month')['price'].sum()

monthly_sales.plot(kind='line', figsize=(10,5), marker='o', color='green')
plt.title('Mothly Sales Trend')
plt.xlabel('Order Month')
plt.ylabel('Price')
plt.show()
```

TOP SELLING PRODUCT CATEGORIES

```python
top_categories = sales.groupby('product_category_name')['price'].sum().sort_values(ascending=False).head(10)

top_categories.plot(kind='bar', figsize=(10,5))
plt.title('Top Product Categories')
plt.xlabel('Product Name')
plt.ylabel('Price')
plt.show()
```

7. PAYMENT METHOD ANALYSIS

```python
payment_counts = sales['payment_type'].value_counts()

payment_counts.plot(kind='pie', autopct='%1.1f%%')
plt.title('Payment Method Distribution')
plt.show()
```

8. DELIVERY PERFORMANCE

```python
sns.histplot(sales['delivery_days'], bins=30)
plt.title('Delivery Time Distribution')
plt.show()
```

9. REVIEW SCORE ANALYSIS

```python
review_dist = sales['review_score'].value_counts()

review_dist.plot(kind='bar')
plt.title('Customer review Score')
plt.show()
```

10. correlation Analysis

```python
corr = sales[['price', 'freight_value', 'delivery_days']].corr()

sns.heatmap(corr, annot=True)
plt.show()
```
